## Demographic Data Analysis

```python
#Imports required libraries
import geopandas as gpd
```

```python
#Imports population shapefile
data_path=r"D:\GIS-TU Dublin\Year_2\THESIS\DCC_Analysis"
shape_file_path= data_path + r"\Data_used\Population_DCC\Population_Small_Areas_DCC.shp"
```

```python
pop_gdf= gpd.read_file(shape_file_path)
```

```python
pop_gdf.head()
```

```python
# List all column names
for col in pop_gdf.columns:
    print(col)
```

```python
# Keep only relevant columns
keep_cols = [ "SA_PUB2022", "GEOGDESC", "LA_CODE", "COUNTY", "AREA","CSO_LEA","LEA_ID",
    "T1_1AGETT", "T1_1AGE65_", "T1_1AGE70_", "T1_1AGE75_", 
    "T1_1AGE80_", "T1_1AGEGE_", "geometry"
]
pop_gdf_filter = pop_gdf[keep_cols]
```

```python
pop_gdf_filter.head()
```

```python
# Rename selected columns to simpler names
rename_map = {
    "SA_PUB2022": "SA_ID",
    "GEOGDESC": "SA_Name",
    "T1_1AGETT": "TotalPop",
    "T1_1AGE65_": "Age65_69",
    "T1_1AGE70_": "Age70_74",
    "T1_1AGE75_": "Age75_79",
    "T1_1AGE80_": "Age80_84",
    "T1_1AGEGE_": "Age85plus",
}
renamed_pop_gdf = pop_gdf_filter.rename(columns=rename_map)
renamed_pop_gdf.head()
```

```python
# Calculate total elderly population (65+)
renamed_pop_gdf["T_65Plus"] = (
    renamed_pop_gdf["Age65_69"] +
    renamed_pop_gdf["Age70_74"] +
    renamed_pop_gdf["Age75_79"] +
    renamed_pop_gdf["Age80_84"] +
    renamed_pop_gdf["Age85plus"]
)

# Calculate percentage of elderly in each Small Area
renamed_pop_gdf["ElderlyPct"] = (renamed_pop_gdf["T_65Plus"] / renamed_pop_gdf["TotalPop"]) * 100
```

```python
renamed_pop_gdf.head()
```

```python
# Save renamed_pop_gdf as a shapefile
renamed_pop_gdf.to_file(data_path + r"\Data_used\Population_DCC\ElderlyPop.shp", driver="ESRI Shapefile", encoding="utf-8")
```
