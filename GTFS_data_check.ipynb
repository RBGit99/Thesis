# Conversion of .txt file to .csv file

```python
#Conversion of .txt file to .csv file


# Define input and output directories
input_dir = r'D:\GIS-TU Dublin\Year_2\Research_Proj_Dissertation\Thesis Proposal\data\data_check\DublinBus_GTFS_txt'
output_dir = r'D:\GIS-TU Dublin\Year_2\Research_Proj_Dissertation\Thesis Proposal\data\data_check\DublinBus_GTFS_csv'

# Create the output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Define the delimiter used in the .txt files
delimiter = ','

# Iterate through all .txt files in the input directory
for filename in os.listdir(input_dir):
    if filename.endswith('.txt'):
        # Define the full paths for input and output files
        input_txt_file = os.path.join(input_dir, filename)
        output_csv_file = os.path.join(output_dir, filename.replace('.txt', '.csv'))

        # Open the .txt file for reading and the .csv file for writing
        try:
            with open(input_txt_file, 'r',encoding='utf-8') as txt_file, open(output_csv_file, 'w', newline='',encoding='utf-8') as csv_file:
                # Create a CSV writer object
                csv_writer = csv.writer(csv_file)
                
                # Read each line from the .txt file
                for line in txt_file:
                    # Split the line into columns based on the delimiter
                    row = line.strip().split(delimiter)
                    
                    # Write the row to the CSV file
                    csv_writer.writerow(row)

            print(f"Converted {filename} to {output_csv_file}")
        except PermissionError:
            print(f"Permission denied: Could not process {filename}.")
        except Exception as e:
            print(f"An error occurred while processing {filename}: {e}")

print("All conversions complete.")
```
